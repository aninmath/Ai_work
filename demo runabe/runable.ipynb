```python
class PromptTemplate:
    
    def __init__(self, input_variable, template):
        self.input_variable = input_variable
        self.template = template

    def invoke (self, prompt : str):

        dummy = [ 'india is a country', 'cricket is a game',
                 'football is best']
        
        return {'response':random.choice(dummy )}

    def format (self, input_dict):

        return self.template.format(**input_dict)
        
```

```python
template = PromptTemplate(
    input_variable= ['name'],
    template= 'write short story about {name}'
)
```

```python
template.format({'name': 'hitler'})
```

```python
class Naklillmchain:

    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def predict (self, dict):
        finalprm = self.prompt.format(dict)
        return (self.llm.predict(finalprm))
```

```python
chain = Naklillmchain(llm, template)
```

```python
chain.predict({'name': 'rabi'})
```
